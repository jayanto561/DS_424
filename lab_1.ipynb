```python
import numpy as np
```

```python
# Sigmoid Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
# Sigmoid Derivative
def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
# Inputs
x1 = 0.6
x2 = 0.8
```

```python
# Weights and Biases
w1 = 0.4
w2 = 0.7
b1 = 0.2

w3 = 0.9
w4 = 0.5

# Target
Y = 1

# Learning Rate
lr = 0.1
```

```python
# Training Loop
for epoch in range(50):

    # Forward Propagation
    z1 = (x1 * w1) + (x2 * w2)+ b1
    h1 = sigmoid(z1)

    z2 = (h1 * w3) 
    h2 = sigmoid(z2)

    z3 = (h2 * w4)
    y1 = sigmoid(z3)


 # Error
    error = Y - y1
# Backpropagation
    delta_y1 = error * sigmoid_derivative(y1)
    delta_h2 = delta_y1 * w4 * sigmoid_derivative(h2)
    delta_h1 = delta_h2 * w3 * sigmoid_derivative(h1)

# Update Output Layer
    w4 = w4 + (lr * h2 * delta_y1)
    w3 = w3 + (lr * h1 * delta_h2)

    # Update Hidden Layer
    w1 = w1 + (lr * x1 * delta_h1)
    w2 = w2 + (lr * x2 * delta_h1)
    b1 = b1 + (lr * delta_h1)

    if epoch % 10 == 0:
        print(f"Epoch {epoch}: Output = {y1:.4f}, Error = {error:.4f}")

print("\n===== FINAL RESULTS =====")
print("Final Output =", y1)
print("Final Error =", error)
```
